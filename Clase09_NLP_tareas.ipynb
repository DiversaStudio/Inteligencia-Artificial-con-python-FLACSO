## 6. Primer modelo: spaCy

**spaCy** es una librería que trae modelos entrenados para el español. El que usamos (`es_core_news_sm`) aprendió de miles de **noticias** a reconocer cómo funcionan las palabras.

```python
nlp = spacy.load("es_core_news_sm")
```

Démosle una frase y miremos qué sabe de cada palabra:

```python
frase = nlp("Organizamos una minga con las vecinas del barrio")

for palabra in frase:
    print(palabra.text, palabra.lemma_, palabra.pos_, palabra.is_stop)
```

**Línea por línea**
- `nlp(...)` le pasa la frase al modelo. El resultado es la frase ya analizada.
- `palabra.lemma_` es el **lema**: la forma de diccionario ("vecinas" → "vecina").
- `palabra.pos_` es la **categoría gramatical**: `VERB` verbo, `NOUN` sustantivo, `DET` artículo, `ADP` preposición.
- `palabra.is_stop` dice si es una **palabra vacía** (`True`) o no (`False`).

Nadie escribió estas reglas a mano: el modelo las aprendió de los textos con los que fue entrenado.

**Para comentar:** miren el lema de "Organizamos". ¿Lo hizo bien? ¿Y qué hizo con "minga"? Un modelo entrenado con noticias de España se equivoca justo en las palabras y formas que menos vio.

## 7. Contar otra vez, ahora con lemas

Ahora recorremos todos los mensajes y guardamos solo los lemas de las palabras que **no** son vacías:

```python
documento = nlp(texto)

lemas = []
for palabra in documento:
    if palabra.is_alpha and not palabra.is_stop:
        lemas.append(palabra.lemma_.lower())

Counter(lemas).most_common(20)
```

**Línea por línea**
- `palabra.is_alpha` es `True` si la palabra tiene solo letras. Así dejamos fuera comas, puntos y números.
- `and not palabra.is_stop` deja fuera las palabras vacías.
- Si se cumplen las dos condiciones, guardamos el lema en minúsculas.

Ahora sí aparecen palabras con contenido: "trabajo", "necesitar", "pasaporte", "visa", "pagar"...

Hagamos la nube otra vez, pero con estos conteos:

```python
nube = WordCloud(width=800, height=400, background_color="white").generate_from_frequencies(Counter(lemas))

plt.imshow(nube)
plt.axis("off")
plt.show()
```

Como vamos a repetir estos pasos, los guardamos en una función. Hace exactamente lo mismo que la celda de arriba:

```python
def lemas_de(texto):
    documento = nlp(texto)
    lemas = []
    for palabra in documento:
        if palabra.is_alpha and not palabra.is_stop:
            lemas.append(palabra.lemma_.lower())
    return lemas
```

## 8. Clasificar por temas con reglas

## 9. Sentimiento con un modelo entrenado

Ahora usamos un modelo **ML** entrenado para una sola tarea: decir si un texto es **positivo** (`POS`), **negativo** (`NEG`) o **neutro** (`NEU`).

Usamos **pysentimiento**, una librería hecha en Argentina para el español. Su modelo de sentimiento se entrenó con unos 5 mil **tuits** en distintas variantes del español.

La primera vez descarga el modelo (unos 500 MB).

```python
from pysentimiento import create_analyzer

analizador = create_analyzer(task="sentiment", lang="es")

analizador.predict("Gracias por la información, ya pude registrarme")
```

El modelo devuelve la etiqueta (`output`) y la probabilidad de cada opción (`probas`). Es la misma idea de la clase 8: el modelo no "sabe", calcula probabilidades.

Para aplicarlo a toda la tabla, hacemos una función pequeña que devuelve solo la etiqueta, y la aplicamos a cada mensaje con `apply`:

```python
def sentimiento(texto):
    return analizador.predict(texto).output

mensajes["sentimiento"] = mensajes["mensaje"].apply(sentimiento)
mensajes[["mensaje", "sentimiento"]]
```
